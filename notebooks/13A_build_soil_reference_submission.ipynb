# 13A — Build Soil Reference Submission

Fast calibration builder. It copies the **actual submitted tarball** from the attached public `Soil Remembers Rain` notebook, validates archive paths and the presence of `main.py`, and writes a fingerprinted output.

Use this only if you want an immediate ladder calibration before running the full V3 transplant lab.

## Input
Attach `Kaggriculture Frontier | The Soil Remembers Rain`.

## Settings
- Accelerator: None
- Internet: OFF


```python
from pathlib import Path
import tarfile,shutil,hashlib,json
INPUT=Path('/kaggle/input')
OUT=Path('/kaggle/working/NEXT_SUBMIT_00_soil_reference.tar.gz')

hits=[]
for p in INPUT.rglob('submission.tar.gz'):
    if 'kaggriculture-frontier-the-soil-remembers-rain' in str(p).lower():
        hits.append(p)
if not hits:
    raise FileNotFoundError('Attach Kaggriculture Frontier | The Soil Remembers Rain')
src=hits[0]
print('source:',src)

with tarfile.open(src,'r:*') as tf:
    files=[]
    for m in tf.getmembers():
        rel=Path(m.name)
        if rel.is_absolute() or '..' in rel.parts:
            raise RuntimeError(f'unsafe archive path: {m.name}')
        if m.isfile(): files.append(m.name)
    if not any(Path(x).name=='main.py' for x in files):
        raise RuntimeError('submission archive has no main.py')
    print('files:',files)

shutil.copy2(src,OUT)
sha=hashlib.sha256(OUT.read_bytes()).hexdigest()
manifest={'source':str(src),'output':str(OUT),'bytes':OUT.stat().st_size,'sha256':sha,'files':files}
Path('/kaggle/working/NEXT_SUBMIT_00_soil_reference.json').write_text(json.dumps(manifest,indent=2))
print(json.dumps(manifest,indent=2))
```
