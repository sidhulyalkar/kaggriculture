# 13 — V3 Frontier Transplant Lab

Self-contained Kaggle CPU experiment for rapid frontier improvement.

It tests pure public frontier references, Soil micro/routing with alternative market policies, reverse controls, and day-7/day-11 phase transplants. All source policies are shadow-called every turn so stateful policies remain initialized.

A hybrid is promoted only if it has zero invalid games, beats pure Soil by at least `+0.03` on held-out robust score, and retains at least 90% of Soil passive cash. Otherwise the notebook outputs the exact Soil reference.
