# V44 Frontier Margin Engine
Qualify the current Moon/Soil frontier, compile a single-parent market residual, run disjoint both-seat screen + held-out gates, and emit a submission **only if it beats the compiled parent robustly**.

Recommended inputs: current Moon, current Soil, exact V32, plus Adaptive/Ranker/Strict/3094/V16/Weed Slip where available.

```python
from pathlib import Path
import json, os, subprocess, sys
INPUT=Path('/kaggle/input')
WORK=Path('/kaggle/working/v44_frontier_margin')
hits=list(INPUT.rglob('scripts/v44_frontier_margin_engine.py'))
if not hits:
    raise FileNotFoundError('Attach the Kaggriculture code-repo dataset containing V44')
hits.sort(key=lambda p:(len(p.parts),str(p)))
REPO=hits[0].parents[1]
print('REPO =',REPO)
print('CPU workers =',min(4,os.cpu_count() or 2))
```

```python
patterns={
 'moon':'kaggriculture-frontier-the-moon-counts-melons',
 'soil':'kaggriculture-frontier-the-soil-remembers-rain',
 'adaptive':'adaptive-farming-strategy-for-kaggriculture',
 'ranker':'kaggriculture-rank-your-agent',
 'strict':'25-27-strict-future-v27-midgame-meta-reset',
 'score3094':'3094-score-kaggriculture',
 'v16':'v16-rc5-high-score-8c-4s-premium-market-lead',
 'weed_slip':'weed-slip',
}
for key,pat in patterns.items():
    found=any(pat in str(p).lower() for p in INPUT.iterdir())
    print(f'{key:10s}', 'attached' if found else '-')
print('V32 archive:', bool(list(INPUT.rglob('SUBMIT_V32_RUNTIME_VERIFIED.tar.gz')) or list(INPUT.rglob('SUBMIT_V32_PREMIUM_FRONT_SINGLEFILE.tar.gz'))))
```

```python
cmd=[sys.executable,str(REPO/'scripts'/'v44_frontier_margin_engine.py'),
     '--input-root',str(INPUT),'--repo',str(REPO),'--work',str(WORK),
     '--workers',str(min(4,os.cpu_count() or 2)),'--screen-seeds','4','--heldout-seeds','8']
print(' '.join(cmd))
subprocess.run(cmd,check=True)
```

```python
decision=json.loads((WORK/'V44_DECISION.json').read_text())
print(json.dumps(decision,indent=2))
print('\nArtifacts:')
for p in sorted(WORK.iterdir()):
    if p.is_file(): print(f'{p.name:38s} {p.stat().st_size:>10,d} bytes')
```
